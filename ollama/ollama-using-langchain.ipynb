#### FOR CHAT CONVERSATION

```python
from langchain_ollama import ChatOllama

# Initialize the model
llm = ChatOllama(model="llama3.2:1b")

# Simple invoke
response = llm.invoke("Write a short poem about the beauty of nature.")
print(response.content)
```

#### EMBEDDING GENERATION USING OLLAMA

```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="embeddinggemma:latest")

# Embed a single query
query_result = embeddings.embed_query("What is the capital of France?")
print("Query Embedding:", query_result)
```

```python
# Emed multiple documents
doc_results = embeddings.embed_documents(["Document 1 text", "Document 2 text"])

print(f"Embedding length: {len(doc_results)}")
```
