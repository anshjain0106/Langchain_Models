```python
# GIVING MULTIMODAL INPUT TO MODEL

import base64
import ollama

image_path = "images.jpg"

with open(image_path, "rb") as f:
    image_bytes = f.read()
image_base64 = base64.b64encode(image_bytes).decode("utf-8")

response = ollama.generate(model="gemma3:4b", images=[image_base64], prompt="Describe the image in detail.")
print(response.response)
```

```python
# PASSING MULTIPLE IMAGES TO THE MODEL

image_paths = ["images.jpg", "images_1.jpg"]

image_base64_list = []
for i in image_paths:
    with open(i, "rb") as f:
        image_bytes = f.read()
        image_base64_list.append(base64.b64encode(image_bytes).decode("utf-8"))

response = ollama.generate(model="gemma3:4b", images=image_base64_list, prompt="Describe the images in detail of each image.")
print(response.response)
```
